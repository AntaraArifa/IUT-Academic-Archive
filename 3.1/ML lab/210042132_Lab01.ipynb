```python
#Task5
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
column_names = ["sepal_length", "sepal_width", "petal_length", "petal_width", "class"]
iris_data = pd.read_csv(url, header=None, names=column_names)

print(iris_data)
```

```python
#Task6
sepal_length =iris_data["sepal_length"]

mean_sepal_length= np.mean(sepal_length)
median_sepal_length = np.median(sepal_length)
std_sepal_length = np.std(sepal_length)

print("Mean : {mean_sepal_length}")
print("Median :{median_sepal_length}")
print("Standard Deviation :{std_sepal_length}" )


print("Summary of dataset :")
print(iris_data.describe(include= 'all'))
```

```python
#Task1

file_path = "sample_data/california_housing_test.csv"
housing_data = pd.read_csv(file_path)

print("First 10 rows of the dataset:")
print(housing_data.head(10))

print("\nData types of each column:")
print(housing_data.dtypes)

print("\nSummary statistics of numerical columns:")
print(housing_data.describe())
```

```python
#Task2

missing_values_count = housing_data.isnull().sum()
print("\nMissing values in each column:")
print(missing_values_count)

housing_data_dropped = housing_data.dropna()
print("\nDataset after removing rows with missing values:")
print(housing_data_dropped)

housing_data_filled = housing_data.copy()
for column in housing_data_filled.columns:
    if np.issubdtype(housing_data_filled[column].dtype, np.number):
        housing_data_filled[column].fillna(housing_data_filled[column].mean(), inplace=True)
    else:
        housing_data_filled[column].fillna(housing_data_filled[column].mode()[0], inplace=True)

print("\nDataset after filling missing values:")
print(housing_data_filled)
```

```python
#Task3

plt.figure(figsize=(10, 6))
plt.hist(housing_data['median_house_value'], bins=50, edgecolor='k')
plt.title('Distribution of Median House Values')
plt.xlabel('Median House Value')
plt.ylabel('Frequency')
plt.show()

top_locations = housing_data.sort_values(by='median_house_value', ascending=False).head(10)
print("\nTop 10 locations with the highest median house values:")
print(top_locations[['longitude', 'latitude', 'median_house_value']])

plt.figure(figsize=(10, 6))
plt.scatter(housing_data['median_income'], housing_data['median_house_value'], alpha=0.5)
plt.title('Median Income vs. Median House Value')
plt.xlabel('Median Income')
plt.ylabel('Median House Value')
plt.show()

print("\nSummary of the top locations for investment:")
print(top_locations.describe())
```
