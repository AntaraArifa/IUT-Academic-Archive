##Task 01

```python
from collections import deque


def count_informed_employees(n, adj, source, time_limit):

    distance = [-1] * (n + 1)
    queue = deque()
    distance[source] = 0
    queue.append(source)

    while queue:
        current = queue.popleft()
        if distance[current] == time_limit:
            continue

        for friend in adj[current]:
            if distance[friend] == -1:
                distance[friend] = distance[current] + 1
                queue.append(friend)

    informed_count = 0

    for employee in range(1, n + 1):
        if 0 <= distance[employee] <= time_limit:
            informed_count += 1

    return informed_count


if __name__ == "__main__":
    n, m, x, t = map(int, input().split())

    adj = [[] for _ in range(n + 1)]

    for _ in range(m):
        u, v = map(int, input().split())

        adj[u].append(v)
        adj[v].append(u)

    answer = count_informed_employees(n, adj, x, t)
    print(answer)
```

##Task 02

```python

def contains_cycle(course, adj, state):

    state[course] = 1

    for next_course in adj[course]:
        if state[next_course] == 0:
            if contains_cycle(next_course, adj, state):
                return True

        elif state[next_course] == 1:
            return True

    state[course] = 2
    return False


if __name__ == "__main__":
    n, m = map(int, input().split())

    adj = [[] for _ in range(n + 1)]

    for _ in range(m):
        u, v = map(int, input().split())

        adj[u].append(v)

    state = [0] * (n + 1)
    cycle_found = False

    for course in range(1, n + 1):
        if state[course] == 0:
            if contains_cycle(course, adj, state):
                cycle_found = True
                break

    if cycle_found:
        print("YES")
    else:
        print("NO")
```
