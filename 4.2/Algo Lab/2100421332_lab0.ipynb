Find a peak element in a 2D matrix.
    A peak is an element >= all 4 neighbors (North, South, East, West).
    
    Algorithm: Divide and Conquer on columns
    - Pick the middle column
    - Find the maximum element in that column
    - Check its left and right neighbors
    - If left neighbor is larger -> recurse on left half
    - If right neighbor is larger -> recurse on right half
    - Otherwise -> current position is a peak
    
    Time Complexity: O(row * log(col))
    Space Complexity: O(log(col)) due to recursion stack

```python
def find_peak_2d(matrix):
    if not matrix or not matrix[0]:
        return None

    rows = len(matrix)
    cols = len(matrix[0])

    def find_max_row_in_col(col):
        """Find the row index of the maximum element in a given column."""
        max_row = 0
        for r in range(1, rows):
            if matrix[r][col] > matrix[max_row][col]:
                max_row = r
        return max_row

    lo, hi = 0, cols - 1

    while lo <= hi:
        mid = (lo + hi) // 2
        max_row = find_max_row_in_col(mid)

        left  = matrix[max_row][mid - 1] if mid > 0        else float('-inf')
        right = matrix[max_row][mid + 1] if mid < cols - 1 else float('-inf')

        if matrix[max_row][mid] >= left and matrix[max_row][mid] >= right:
            return (max_row, mid)
        elif left > matrix[max_row][mid]:
            hi = mid - 1
        else:
            lo = mid + 1

    return None
```

Helper: verify a found peak is truly valid

```python
def is_valid_peak(matrix, r, c):
    val = matrix[r][c]
    rows, cols = len(matrix), len(matrix[0])
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols:
            if matrix[nr][nc] > val:
                return False
    return True


def run_test(name, matrix):
    print(f"\n{'='*45}")
    print(f"Test: {name}")
    for row in matrix:
        print(" ", row)
    result = find_peak_2d(matrix)
    if result is None:
        print("Result: No peak found.")
    else:
        r, c = result
        valid = is_valid_peak(matrix, r, c)
        print(f"Peak found at: ({r}, {c})  ->  value = {matrix[r][c]}")
        print(f"Valid peak:    {valid}")
```
