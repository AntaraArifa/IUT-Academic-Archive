# TASK 1
# Next Greater Element

```python
def next_greater_element(arr):
    n = len(arr)
    result = [-1] * n
    stack = []

    for i in range(n - 1, -1, -1):
        while stack and stack[-1] <= arr[i]:
            stack.pop()

        if stack:
            result[i] = stack[-1]

        stack.append(arr[i])

    return result


print("TASK 1 OUTPUT")

arr1 = [1, 3, 2, 4]
print("Input:", arr1)
print("Output:", next_greater_element(arr1))

arr2 = [6, 8, 0, 1, 3]
print("Input:", arr2)
print("Output:", next_greater_element(arr2))
```

# TASK 2
# Stack using Two Queues

```python
from collections import deque

class StackUsingTwoQueues:
    def __init__(self):
        self.q1 = deque()
        self.q2 = deque()

    def push(self, x):
        self.q2.append(x)

        while self.q1:
            self.q2.append(self.q1.popleft())

        self.q1, self.q2 = self.q2, self.q1

    def pop(self):
        if not self.q1:
            return None
        return self.q1.popleft()

    def top(self):
        if not self.q1:
            return None
        return self.q1[0]

    def empty(self):
        return len(self.q1) == 0

    def size(self):
        return len(self.q1)


print("\nTASK 2 OUTPUT")

st = StackUsingTwoQueues()

st.push(10)
st.push(20)
st.push(30)

print("After pushing 10, 20, 30")
print("Pop:", st.pop())
print("Pop:", st.pop())
print("Top:", st.top())
print("Size:", st.size())
print("Empty:", st.empty())
print("Pop:", st.pop())
print("Empty:", st.empty())
```

# TASK 3
# First Non-Repeating Character in Stream

```python
def first_non_repeating(s):
    freq = {}
    q = deque()
    result = ""

    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1
        q.append(ch)

        while q and freq[q[0]] > 1:
            q.popleft()

        if q:
            result += q[0]
        else:
            result += "#"

    return result


print("\nTASK 3 OUTPUT")

s1 = "aabc"
print("Input:", s1)
print("Output:", first_non_repeating(s1))

s2 = "bb"
print("Input:", s2)
print("Output:", first_non_repeating(s2))
```
